```python
"""
ДЗ 13. Управление системами и ресурсами защиты информации
с помощью Python

Описание:
Скрипт выполняет запрос к API VirusTotal по SHA256-хэшу файла
и выводит статистику анализа (malicious, suspicious, harmless, undetected).

Перед запуском необходимо:

1. Установить библиотеку requests:
   pip install requests

2. Задать переменную окружения с API-ключом:

   Linux / Mac:
   export VT_API_KEY="ваш_ключ"

   Windows:
   setx VT_API_KEY "ваш_ключ"

3. Запустить:
   python vt_api.py
"""

import os
import requests


def get_file_report(api_key: str, file_hash: str) -> dict:
    """
    Отправляет GET-запрос к VirusTotal API
    и возвращает JSON-ответ.
    """
    url = f"https://www.virustotal.com/api/v3/files/{file_hash}"
    headers = {"x-apikey": api_key}

    response = requests.get(url, headers=headers, timeout=15)

    if response.status_code != 200:
        raise Exception(
            f"Ошибка запроса: {response.status_code}\n{response.text}"
        )

    return response.json()


def print_stats(data: dict) -> None:
    """
    Извлекает из JSON статистику анализа файла
    и выводит её в консоль.
    """
    stats = data["data"]["attributes"]["last_analysis_stats"]

    print("Статистика сканирования файла:")
    print(f"malicious: {stats.get('malicious', 0)}")
    print(f"suspicious: {stats.get('suspicious', 0)}")
    print(f"harmless: {stats.get('harmless', 0)}")
    print(f"undetected: {stats.get('undetected', 0)}")


def main():
    # Получаем API-ключ из переменной окружения
    API_KEY = "ваш_ключ"

    # Тестовый SHA256-хэш (EICAR)
    FILE_HASH = "44d88612fea8a8f36de82e1278abb02f"

    if not API_KEY:
        raise ValueError(
            "API ключ не найден. Задайте переменную окружения VT_API_KEY."
        )

    print("Отправка запроса к VirusTotal...\n")

    data = get_file_report(API_KEY, FILE_HASH)

    print("JSON получен успешно.\n")

    # Корректный вывод обработанных данных
    print_stats(data)


if __name__ == "__main__":
    main()
```
