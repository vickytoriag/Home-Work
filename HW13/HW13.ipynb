```python
import os
import json
import requests

API_KEY = os.getenv("VT_API_KEY")

FILE_HASH = "44d88612fea8a8f36de82e1278abb02f"

if not API_KEY:
    raise ValueError("API ключ не найден.")

url = f"https://www.virustotal.com/api/v3/files/{FILE_HASH}"
headers = {"x-apikey": API_KEY}

response = requests.get(url, headers=headers, timeout=30)

if response.status_code != 200:
    print("Ошибка запроса:", response.status_code)
    print(response.text)
    raise SystemExit(1)

data = response.json()

print("Запрос выполнен успешно. JSON ответ получен.\n")

# 1) Печать JSON в консоль
print(json.dumps(data, indent=2, ensure_ascii=False))

# 2) Сохранение JSON в файл
out_file = "vt_response.json"
with open(out_file, "w", encoding="utf-8") as f:
    json.dump(data, f, indent=2, ensure_ascii=False)

print(f"\nJSON сохранен в файл: {out_file}")

# 3) Короткий вывод статистики (если поле есть)
try:
    stats = data["data"]["attributes"]["last_analysis_stats"]
    print("\nСтатистика сканирования файла:")
    print("malicious:", stats.get("malicious"))
    print("suspicious:", stats.get("suspicious"))
    print("harmless:", stats.get("harmless"))
    print("undetected:", stats.get("undetected"))
except Exception:
    print("\nСтатистику last_analysis_stats не удалось извлечь из ответа.")
```
