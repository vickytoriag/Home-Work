```python
from pathlib import Path
import ipywidgets as widgets
from IPython.display import display, clear_output

S_BOX = [
    0x63, 0x7c, 0x77, 0x7b, 0xf2, 0x6b, 0x6f, 0xc5, 0x30, 0x01, 0x67, 0x2b, 0xfe, 0xd7, 0xab, 0x76,
    0xca, 0x82, 0xc9, 0x7d, 0xfa, 0x59, 0x47, 0xf0, 0xad, 0xd4, 0xa2, 0xaf, 0x9c, 0xa4, 0x72, 0xc0,
    0xb7, 0xfd, 0x93, 0x26, 0x36, 0x3f, 0xf7, 0xcc, 0x34, 0xa5, 0xe5, 0xf1, 0x71, 0xd8, 0x31, 0x15,
    0x04, 0xc7, 0x23, 0xc3, 0x18, 0x96, 0x05, 0x9a, 0x07, 0x12, 0x80, 0xe2, 0xeb, 0x27, 0xb2, 0x75,
    0x09, 0x83, 0x2c, 0x1a, 0x1b, 0x6e, 0x5a, 0xa0, 0x52, 0x3b, 0xd6, 0xb3, 0x29, 0xe3, 0x2f, 0x84,
    0x53, 0xd1, 0x00, 0xed, 0x20, 0xfc, 0xb1, 0x5b, 0x6a, 0xcb, 0xbe, 0x39, 0x4a, 0x4c, 0x58, 0xcf,
    0xd0, 0xef, 0xaa, 0xfb, 0x43, 0x4d, 0x33, 0x85, 0x45, 0xf9, 0x02, 0x7f, 0x50, 0x3c, 0x9f, 0xa8,
    0x51, 0xa3, 0x40, 0x8f, 0x92, 0x9d, 0x38, 0xf5, 0xbc, 0xb6, 0xda, 0x21, 0x10, 0xff, 0xf3, 0xd2,
    0xcd, 0x0c, 0x13, 0xec, 0x5f, 0x97, 0x44, 0x17, 0xc4, 0xa7, 0x7e, 0x3d, 0x64, 0x5d, 0x19, 0x73,
    0x60, 0x81, 0x4f, 0xdc, 0x22, 0x2a, 0x90, 0x88, 0x46, 0xee, 0xb8, 0x14, 0xde, 0x5e, 0x0b, 0xdb,
    0xe0, 0x32, 0x3a, 0x0a, 0x49, 0x06, 0x24, 0x5c, 0xc2, 0xd3, 0xac, 0x62, 0x91, 0x95, 0xe4, 0x79,
    0xe7, 0xc8, 0x37, 0x6d, 0x8d, 0xd5, 0x4e, 0xa9, 0x6c, 0x56, 0xf4, 0xea, 0x65, 0x7a, 0xae, 0x08,
    0xba, 0x78, 0x25, 0x2e, 0x1c, 0xa6, 0xb4, 0xc6, 0xe8, 0xdd, 0x74, 0x1f, 0x4b, 0xbd, 0x8b, 0x8a,
    0x70, 0x3e, 0xb5, 0x66, 0x48, 0x03, 0xf6, 0x0e, 0x61, 0x35, 0x57, 0xb9, 0x86, 0xc1, 0x1d, 0x9e,
    0xe1, 0xf8, 0x98, 0x11, 0x69, 0xd9, 0x8e, 0x94, 0x9b, 0x1e, 0x87, 0xe9, 0xce, 0x55, 0x28, 0xdf,
    0x8c, 0xa1, 0x89, 0x0d, 0xbf, 0xe6, 0x42, 0x68, 0x41, 0x99, 0x2d, 0x0f, 0xb0, 0x54, 0xbb, 0x16
]

INV_S_BOX = [
    0x52, 0x09, 0x6a, 0xd5, 0x30, 0x36, 0xa5, 0x38, 0xbf, 0x40, 0xa3, 0x9e, 0x81, 0xf3, 0xd7, 0xfb,
    0x7c, 0xe3, 0x39, 0x82, 0x9b, 0x2f, 0xff, 0x87, 0x34, 0x8e, 0x43, 0x44, 0xc4, 0xde, 0xe9, 0xcb,
    0x54, 0x7b, 0x94, 0x32, 0xa6, 0xc2, 0x23, 0x3d, 0xee, 0x4c, 0x95, 0x0b, 0x42, 0xfa, 0xc3, 0x4e,
    0x08, 0x2e, 0xa1, 0x66, 0x28, 0xd9, 0x24, 0xb2, 0x76, 0x5b, 0xa2, 0x49, 0x6d, 0x8b, 0xd1, 0x25,
    0x72, 0xf8, 0xf6, 0x64, 0x86, 0x68, 0x98, 0x16, 0xd4, 0xa4, 0x5c, 0xcc, 0x5d, 0x65, 0xb6, 0x92,
    0x6c, 0x70, 0x48, 0x50, 0xfd, 0xed, 0xb9, 0xda, 0x5e, 0x15, 0x46, 0x57, 0xa7, 0x8d, 0x9d, 0x84,
    0x90, 0xd8, 0xab, 0x00, 0x8c, 0xbc, 0xd3, 0x0a, 0xf7, 0xe4, 0x58, 0x05, 0xb8, 0xb3, 0x45, 0x06,
    0xd0, 0x2c, 0x1e, 0x8f, 0xca, 0x3f, 0x0f, 0x02, 0xc1, 0xaf, 0xbd, 0x03, 0x01, 0x13, 0x8a, 0x6b,
    0x3a, 0x91, 0x11, 0x41, 0x4f, 0x67, 0xdc, 0xea, 0x97, 0xf2, 0xcf, 0xce, 0xf0, 0xb4, 0xe6, 0x73,
    0x96, 0xac, 0x74, 0x22, 0xe7, 0xad, 0x35, 0x85, 0xe2, 0xf9, 0x37, 0xe8, 0x1c, 0x75, 0xdf, 0x6e,
    0x47, 0xf1, 0x1a, 0x71, 0x1d, 0x29, 0xc5, 0x89, 0x6f, 0xb7, 0x62, 0x0e, 0xaa, 0x18, 0xbe, 0x1b,
    0xfc, 0x56, 0x3e, 0x4b, 0xc6, 0xd2, 0x79, 0x20, 0x9a, 0xdb, 0xc0, 0xfe, 0x78, 0xcd, 0x5a, 0xf4,
    0x1f, 0xdd, 0xa8, 0x33, 0x88, 0x07, 0xc7, 0x31, 0xb1, 0x12, 0x10, 0x59, 0x27, 0x80, 0xec, 0x5f,
    0x60, 0x51, 0x7f, 0xa9, 0x19, 0xb5, 0x4a, 0x0d, 0x2d, 0xe5, 0x7a, 0x9f, 0x93, 0xc9, 0x9c, 0xef,
    0xa0, 0xe0, 0x3b, 0x4d, 0xae, 0x2a, 0xf5, 0xb0, 0xc8, 0xeb, 0xbb, 0x3c, 0x83, 0x53, 0x99, 0x61,
    0x17, 0x2b, 0x04, 0x7e, 0xba, 0x77, 0xd6, 0x26, 0xe1, 0x69, 0x14, 0x63, 0x55, 0x21, 0x0c, 0x7d
]

RCON = [0x01, 0x02, 0x04, 0x08, 0x10, 0x20, 0x40, 0x80, 0x1B, 0x36]


def xtime(a: int) -> int:
    if a & 0x80:
        return ((a << 1) ^ 0x1B) & 0xFF
    return (a << 1) & 0xFF


def gf_mul(a: int, b: int) -> int:
    p = 0
    for _ in range(8):
        if b & 1:
            p ^= a
        hi = a & 0x80
        a = (a << 1) & 0xFF
        if hi:
            a ^= 0x1B
        b >>= 1
    return p


class AESCipher:
    def __init__(self, key: bytes):
        if len(key) != 16:
            raise ValueError("Ключ должен быть ровно 16 байт для AES-128")
        self.key = key
        self.round_keys = self.key_expansion(key)

    def key_expansion(self, key: bytes):
        words = [list(key[i * 4:(i + 1) * 4]) for i in range(4)]

        for i in range(4, 44):
            temp = words[i - 1][:]
            if i % 4 == 0:
                temp = temp[1:] + temp[:1]
                temp = [S_BOX[b] for b in temp]
                temp[0] ^= RCON[i // 4 - 1]
            new_word = [words[i - 4][j] ^ temp[j] for j in range(4)]
            words.append(new_word)

        round_keys = []
        for r in range(11):
            key_block = []
            for j in range(4):
                key_block.extend(words[r * 4 + j])
            round_keys.append(key_block)

        return round_keys

    def sub_bytes(self, state, inv=False):
        box = INV_S_BOX if inv else S_BOX
        return [box[x] for x in state]

    def shift_rows(self, state, inv=False):
        s = state[:]
        if not inv:
            s[1], s[5], s[9], s[13] = state[5], state[9], state[13], state[1]
            s[2], s[6], s[10], s[14] = state[10], state[14], state[2], state[6]
            s[3], s[7], s[11], s[15] = state[15], state[3], state[7], state[11]
        else:
            s[1], s[5], s[9], s[13] = state[13], state[1], state[5], state[9]
            s[2], s[6], s[10], s[14] = state[10], state[14], state[2], state[6]
            s[3], s[7], s[11], s[15] = state[7], state[11], state[15], state[3]
        return s

    def mix_columns(self, state, inv=False):
        result = state[:]
        for i in range(4):
            col = state[i * 4:(i + 1) * 4]
            if not inv:
                result[i * 4] = xtime(col[0]) ^ (xtime(col[1]) ^ col[1]) ^ col[2] ^ col[3]
                result[i * 4 + 1] = col[0] ^ xtime(col[1]) ^ (xtime(col[2]) ^ col[2]) ^ col[3]
                result[i * 4 + 2] = col[0] ^ col[1] ^ xtime(col[2]) ^ (xtime(col[3]) ^ col[3])
                result[i * 4 + 3] = (xtime(col[0]) ^ col[0]) ^ col[1] ^ col[2] ^ xtime(col[3])
            else:
                result[i * 4] = gf_mul(col[0], 0x0E) ^ gf_mul(col[1], 0x0B) ^ gf_mul(col[2], 0x0D) ^ gf_mul(col[3], 0x09)
                result[i * 4 + 1] = gf_mul(col[0], 0x09) ^ gf_mul(col[1], 0x0E) ^ gf_mul(col[2], 0x0B) ^ gf_mul(col[3], 0x0D)
                result[i * 4 + 2] = gf_mul(col[0], 0x0D) ^ gf_mul(col[1], 0x09) ^ gf_mul(col[2], 0x0E) ^ gf_mul(col[3], 0x0B)
                result[i * 4 + 3] = gf_mul(col[0], 0x0B) ^ gf_mul(col[1], 0x0D) ^ gf_mul(col[2], 0x09) ^ gf_mul(col[3], 0x0E)
        return result

    def add_round_key(self, state, round_key):
        return [state[i] ^ round_key[i] for i in range(16)]

    def encrypt_block(self, block: bytes) -> bytes:
        if len(block) != 16:
            raise ValueError("Блок должен быть длиной 16 байт")

        state = list(block)
        state = self.add_round_key(state, self.round_keys[0])

        for r in range(1, 10):
            state = self.sub_bytes(state)
            state = self.shift_rows(state)
            state = self.mix_columns(state)
            state = self.add_round_key(state, self.round_keys[r])

        state = self.sub_bytes(state)
        state = self.shift_rows(state)
        state = self.add_round_key(state, self.round_keys[10])

        return bytes(state)

    def decrypt_block(self, block: bytes) -> bytes:
        if len(block) != 16:
            raise ValueError("Блок должен быть длиной 16 байт")

        state = list(block)
        state = self.add_round_key(state, self.round_keys[10])

        for r in range(9, 0, -1):
            state = self.shift_rows(state, inv=True)
            state = self.sub_bytes(state, inv=True)
            state = self.add_round_key(state, self.round_keys[r])
            state = self.mix_columns(state, inv=True)

        state = self.shift_rows(state, inv=True)
        state = self.sub_bytes(state, inv=True)
        state = self.add_round_key(state, self.round_keys[0])

        return bytes(state)

    def pad(self, data: bytes) -> bytes:
        pad_len = 16 - (len(data) % 16)
        return data + bytes([pad_len] * pad_len)

    def unpad(self, data: bytes) -> bytes:
        if len(data) == 0:
            raise ValueError("Пустые данные")

        pad_len = data[-1]
        if pad_len < 1 or pad_len > 16:
            raise ValueError("Некорректное дополнение")

        if data[-pad_len:] != bytes([pad_len] * pad_len):
            raise ValueError("Некорректное дополнение")

        return data[:-pad_len]

    def encrypt_data(self, data: bytes) -> bytes:
        data = self.pad(data)
        result = bytearray()

        for i in range(0, len(data), 16):
            result.extend(self.encrypt_block(data[i:i + 16]))

        return bytes(result)

    def decrypt_data(self, data: bytes) -> bytes:
        if len(data) % 16 != 0:
            raise ValueError("Длина шифртекста должна быть кратна 16 байтам")

        result = bytearray()
        for i in range(0, len(data), 16):
            result.extend(self.decrypt_block(data[i:i + 16]))

        return self.unpad(bytes(result))

    def process_file(self, input_path: str, output_path: str, mode="encrypt"):
        with open(input_path, "rb") as f:
            data = f.read()

        if mode == "encrypt":
            result = self.encrypt_data(data)
        else:
            result = self.decrypt_data(data)

        with open(output_path, "wb") as f:
            f.write(result)

        return {
            "input_size": len(data),
            "output_size": len(result),
            "output_path": output_path
        }


TEST_TEXT = "Your brain requires rest."
DEFAULT_KEY = "SecureKey2026!!!"


title = widgets.HTML(
    "<h2 style='margin:0 0 8px 0;'>AES-128</h2>"
    "<p style='margin:0 0 16px 0;'>Учебная программа шифрования и расшифрования</p>"
)

key_input = widgets.Text(
    value=DEFAULT_KEY,
    description="Ключ:",
    style={"description_width": "110px"},
    layout=widgets.Layout(width="700px")
)

plain_input = widgets.Textarea(
    value=TEST_TEXT,
    description="Текст:",
    style={"description_width": "110px"},
    layout=widgets.Layout(width="1000px", height="120px")
)

cipher_output = widgets.Textarea(
    value="",
    description="HEX:",
    style={"description_width": "110px"},
    layout=widgets.Layout(width="1000px", height="120px")
)

decrypt_output = widgets.Textarea(
    value="",
    description="Результат:",
    style={"description_width": "110px"},
    layout=widgets.Layout(width="1000px", height="120px")
)

encrypt_btn = widgets.Button(description="Зашифровать текст", button_style="primary", layout=widgets.Layout(width="180px"))
decrypt_btn = widgets.Button(description="Расшифровать HEX", button_style="success", layout=widgets.Layout(width="180px"))
test_btn = widgets.Button(description="Тест строки", button_style="info", layout=widgets.Layout(width="150px"))
nist_btn = widgets.Button(description="NIST-тест", button_style="warning", layout=widgets.Layout(width="150px"))

file_input_name = widgets.Text(
    value="plain_text.txt",
    description="Входной файл:",
    style={"description_width": "110px"},
    layout=widgets.Layout(width="700px")
)

file_enc_name = widgets.Text(
    value="cipher_text.bin",
    description="Шифр файл:",
    style={"description_width": "110px"},
    layout=widgets.Layout(width="700px")
)

file_dec_name = widgets.Text(
    value="restored_text.txt",
    description="Выходной файл:",
    style={"description_width": "110px"},
    layout=widgets.Layout(width="700px")
)

save_plain_btn = widgets.Button(description="Сохранить тестовый файл", layout=widgets.Layout(width="200px"))
encrypt_file_btn = widgets.Button(description="Зашифровать файл", layout=widgets.Layout(width="170px"))
decrypt_file_btn = widgets.Button(description="Расшифровать файл", layout=widgets.Layout(width="170px"))

status_out = widgets.Output(
    layout=widgets.Layout(
        width="1000px",
        min_height="120px",
        border="1px solid #cccccc",
        padding="10px",
        margin="10px 0 0 0"
    )
)


def get_cipher():
    key_text = key_input.value
    try:
        key = key_text.encode("ascii")
    except UnicodeEncodeError:
        raise ValueError("Ключ должен содержать только ASCII-символы.")

    if len(key) != 16:
        raise ValueError(f"Ключ должен быть длиной 16 байт. Сейчас: {len(key)}")

    return AESCipher(key)


def on_encrypt_click(_):
    with status_out:
        clear_output()
        try:
            cipher = get_cipher()
            data = plain_input.value.encode("ascii")
            encrypted = cipher.encrypt_data(data)
            cipher_output.value = encrypted.hex()

            print("Текст зашифрован.")
            print(f"Исходный текст: {plain_input.value}")
            print(f"Размер открытых данных: {len(data)} байт")
            print(f"Размер шифртекста: {len(encrypted)} байт")
        except UnicodeEncodeError:
            print("Ошибка: используйте ASCII-текст.")
        except Exception as e:
            print("Ошибка:", e)


def on_decrypt_click(_):
    with status_out:
        clear_output()
        try:
            cipher = get_cipher()
            hex_data = cipher_output.value.strip().replace(" ", "").replace("\n", "")
            encrypted = bytes.fromhex(hex_data)
            decrypted = cipher.decrypt_data(encrypted)
            decrypt_output.value = decrypted.decode("ascii")
            print("Расшифрование выполнено.")
        except Exception as e:
            print("Ошибка:", e)


def on_test_click(_):
    with status_out:
        clear_output()
        try:
            cipher = get_cipher()
            source = TEST_TEXT.encode("ascii")
            encrypted = cipher.encrypt_data(source)
            decrypted = cipher.decrypt_data(encrypted)

            cipher_output.value = encrypted.hex()
            decrypt_output.value = decrypted.decode("ascii")

            print("ПРОВЕРКА ТЕСТОВОЙ СТРОКИ")
            print("Исходный текст:", TEST_TEXT)
            print("Длина текста:", len(source), "байт")
            print("Шифртекст HEX:", encrypted.hex())
            print("Расшифровка:", decrypted.decode("ascii"))
            print("Совпадение:", decrypted.decode("ascii") == TEST_TEXT)
        except Exception as e:
            print("Ошибка:", e)


def on_nist_click(_):
    with status_out:
        clear_output()
        try:
            key = bytes.fromhex("000102030405060708090a0b0c0d0e0f")
            plaintext = bytes.fromhex("00112233445566778899aabbccddeeff")
            expected = bytes.fromhex("69c4e0d86a7b0430d8cdb78070b4c55a")

            cipher = AESCipher(key)
            result = cipher.encrypt_block(plaintext)

            print("ПРОВЕРКА ПО ТЕСТОВОМУ ВЕКТОРУ NIST")
            print("Ключ:", key.hex())
            print("Открытый текст:", plaintext.hex())
            print("Ожидаемый шифртекст :", expected.hex())
            print("Полученный шифртекст:", result.hex())
            print("Совпадение:", result == expected)
        except Exception as e:
            print("Ошибка:", e)


def on_save_plain_click(_):
    with status_out:
        clear_output()
        try:
            Path(file_input_name.value).write_bytes(TEST_TEXT.encode("ascii"))
            print(f"Файл создан: {file_input_name.value}")
            print(f"Содержимое: {TEST_TEXT}")
        except Exception as e:
            print("Ошибка:", e)


def on_encrypt_file_click(_):
    with status_out:
        clear_output()
        try:
            cipher = get_cipher()
            info = cipher.process_file(file_input_name.value, file_enc_name.value, mode="encrypt")
            print("Файл зашифрован.")
            print(info)
        except Exception as e:
            print("Ошибка:", e)


def on_decrypt_file_click(_):
    with status_out:
        clear_output()
        try:
            cipher = get_cipher()
            info = cipher.process_file(file_enc_name.value, file_dec_name.value, mode="decrypt")
            restored_bytes = Path(file_dec_name.value).read_bytes()

            print("Файл расшифрован.")
            print(info)
            print("Размер восстановленного файла:", len(restored_bytes), "байт")

            if Path(file_input_name.value).exists():
                source_bytes = Path(file_input_name.value).read_bytes()
                print("Совпадение с исходным файлом:", restored_bytes == source_bytes)
        except Exception as e:
            print("Ошибка:", e)


encrypt_btn.on_click(on_encrypt_click)
decrypt_btn.on_click(on_decrypt_click)
test_btn.on_click(on_test_click)
nist_btn.on_click(on_nist_click)
save_plain_btn.on_click(on_save_plain_click)
encrypt_file_btn.on_click(on_encrypt_file_click)
decrypt_file_btn.on_click(on_decrypt_file_click)

buttons_row = widgets.HBox(
    [encrypt_btn, decrypt_btn, test_btn, nist_btn],
    layout=widgets.Layout(gap="10px", margin="10px 0 10px 110px")
)

file_buttons_row = widgets.HBox(
    [save_plain_btn, encrypt_file_btn, decrypt_file_btn],
    layout=widgets.Layout(gap="10px", margin="10px 0 10px 110px")
)

ui = widgets.VBox([
    title,
    key_input,
    plain_input,
    buttons_row,
    cipher_output,
    decrypt_output,
    widgets.HTML("<hr><h3>Работа с файлами</h3>"),
    file_input_name,
    file_enc_name,
    file_dec_name,
    file_buttons_row,
    status_out
])

display(ui)
```
